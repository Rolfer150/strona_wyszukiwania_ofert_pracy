```python
from fastapi import FastAPI
import random

app = FastAPI()

# Symulacja bazy pytań
QUESTIONS_DB = {
    "Backend Developer": [
        "Jak działa garbage collector w Pythonie?",
        "Co to jest ORM?",
        "Różnice między SQL a NoSQL?",
    ],
    "Frontend Developer": [
        "Jak działa Virtual DOM?",
        "Czym różni się React od Vue?",
        "Jak działa flexbox i grid?",
    ]
}

@app.get("/generate_questions")
def generate_questions(category: str):
    questions = QUESTIONS_DB.get(category, ["Brak dostępnych pytań"])
    return {"questions": random.sample(questions, min(3, len(questions)))}  # Wybiera losowe pytania
```
